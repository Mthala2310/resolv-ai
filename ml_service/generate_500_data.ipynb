```python
import requests
import random
import time

URL = "http://localhost:3000/keluhan"

# 1. Komponen untuk membuat variasi Judul dan Deskripsi secara acak
KATEGORI_MASALAH = {
    "Aplikasi & Akun": {
        "judul": ["Gagal login aplikasi", "Aplikasi force close", "Akun terblokir tiba-tiba", "Tidak bisa ganti password", "Error saat verifikasi OTP"],
        "detail": [
            "Saya sudah mencoba masuk berkali-kali menggunakan email resmi tetapi selalu muncul error jaringan.",
            "Setiap kali membuka menu utama, aplikasi langsung menutup sendiri secara tiba-tiba tanpa pesan.",
            "Sistem mengatakan akun saya ditangguhkan padahal saya tidak melakukan pelanggaran apapun mohon dicek.",
            "Kode verifikasi yang dikirim ke nomor WhatsApp tidak pernah masuk padahal sinyal sangat penuh.",
            "Gagal terus saat mengunggah foto KTP untuk verifikasi akun, mohon bantuannya sistem eror terus."
        ]
    },
    "Transaksi & Pembayaran": {
        "judul": ["Tagihan ganda terpotong", "Metode QRIS hilang", "Transfer bank belum masuk", "Gagal checkout belanjaan", "Saldo e-wallet berkurang otomatis"],
        "detail": [
            "Saldo rekening saya terpotong dua kali untuk transaksi yang sama pada pagi hari ini pukul 08.00.",
            "Saat masuk ke halaman pembayaran, opsi untuk menggunakan metode QRIS mendadak hilang tidak bisa dipilih.",
            "Saya sudah transfer sejak 3 jam yang lalu tapi status di aplikasi masih menunggu pembayaran terus.",
            "Muncul pesan error transaksi gagal saat menekan tombol bayar padahal saldo saya sangat mencukupi.",
            "Ada riwayat transaksi misterius yang memotong saldo saya tanpa ada konfirmasi OTP atau notifikasi."
        ]
    },
    "Layanan Pelanggan": {
        "judul": ["Respon CS sangat lambat", "Tiket bantuan ditutup sepihak", "Petugas kurir tidak ramah", "Salah input data oleh admin", "Komplain tidak ditanggapi"],
        "detail": [
            "Saya sudah menunggu balasan dari live chat selama lebih dari 5 jam tapi hanya direspon oleh bot.",
            "Masalah saya belum selesai sama sekali tetapi tiket laporan di aplikasi sudah diubah menjadi status solved.",
            "Paket dikirim ke alamat yang salah dan ketika dihubungi pihak kurir malah marah-marah tidak jelas.",
            "Admin salah memasukkan nomor resi pengiriman sehingga saya tidak bisa melacak keberadaan paket saya.",
            "Sudah kirim email komplain sejak minggu lalu sampai sekarang tidak ada satu pun balasan dari tim terkait."
        ]
    }
}

# 2. Penambahan teks pelengkap agar karakter selalu unik dan panjang kalimat bervariasi
TEKS_TAMBAHAN = [
    "Tolong segera diperbaiki karena ini sangat mengganggu kenyamanan pengguna.",
    "Mohon bantuannya untuk dicek kembali oleh tim IT server kalian.",
    "Saya sangat kecewa dengan pembaruan sistem yang sekarang, jadi banyak kendala.",
    "Ditunggu respon dan solusinya secepat mungkin, terima kasih banyak.",
    "Sudah dicoba install ulang aplikasinya tapi tetap saja hasilnya sama.",
    "Kejadian ini sudah berulang kali terjadi dalam minggu ini.",
    "Saya butuh dana ini segera dikembalikan karena nominalnya cukup besar bagi saya."
]

def buat_payload_unik(id_angka):
    # Pilih kategori secara acak
    kategori_terpilih = random.choice(list(KATEGORI_MASALAH.keys()))
    data_kat = KATEGORI_MASALAH[kategori_terpilih]
    
    # Pilih template judul dan detail acak
    judul_dasar = random.choice(data_kat["judul"])
    detail_dasar = random.choice(data_kat["detail"])
    pelengkap = random.choice(TEKS_TAMBAHAN)
    
    # Gabungkan menjadi satu kesatuan data unik
    judul_final = f"{judul_dasar} (Kasus #{id_angka})"
    deskripsi_final = f"{detail_dasar} {pelengkap} [ID-{id_angka}]"
    
    return {
        "judul": judul_final,
        "deskripsi": deskripsi_final
    }

def kirim_500_data():
    total_data = 500
    sukses = 0
    gagal = 0
    
    print(f"=== Memulai Inject 500 Data Dummy ke {URL} ===")
    
    for i in range(1, total_data + 1):
        payload = buat_payload_unik(i)
        
        try:
            response = requests.post(URL, json=payload, headers={"Content-Type": "application/json"})
            
            # BAGIAN UTAMA YANG SUDAH DIPERBAIKI TOTAL
            if response.status_code in [200, 201]:
                sukses += 1
                print(f"[{i}/{total_data}] Berhasil dikirim -> {payload['judul']}")
            else:
                gagal += 1
                print(f"[{i}/{total_data}] GAGAL. Status: {response.status_code}")
                
        except requests.exceptions.ConnectionError:
            print("\n[CRITICAL ERROR] Koneksi terputus! Pastikan server localhost:3000 Anda tetap menyala.")
            break
            
        # Jeda mikro (0.05 detik) agar proses cepat selesai tapi tidak merusak/overload server lokal
        time.sleep(0.05)
        
    print(f"\n=== PROSES SELESAI ===")
    print(f"Total Berhasil: {sukses}")
    print(f"Total Gagal: {gagal}")

if __name__ == "__main__":
    kirim_500_data()
```
